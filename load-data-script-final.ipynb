```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import pathlib

from scripts import load_data as ld

curr_dir = pathlib.Path(".")
data_dir = curr_dir / "RTS-GMLC-master" / "RTS_Data"
load_data_dir = curr_dir / "load-data"
```

```python
# Initialize load model
load_model = ld.LoadModel()
```

```python
# Train load model on historical data
load_model.train_model(ld.training_buses, ld.training_years, save=True)
```

```python
# Get load profiles of all historical years (2000-2019) for all buses
df_bus = pd.read_csv(curr_dir / "RTS-GMLC-master" / "RTS_Data" / "SourceData" / "bus.csv", index_col=[0])

buses = list(df_bus.index)
years = np.arange(1999, 2020)

load_profiles = load_model.get_load_profiles(buses, years)
```

```python
# Save profiles
for bus in buses:
    df_load_bus = load_profiles[bus]
    bus_dir = load_data_dir / "profiles" / f"bus{bus}"
    bus_dir.mkdir(exist_ok=True, parents=True)
    for year in years:
        df_load_bus_year = df_load_bus.loc[df_load_bus.index.year == year]
        df_load_bus_year.to_csv(bus_dir / f"NSRDB_load-profile_bus{bus}_{year}.csv")
```
